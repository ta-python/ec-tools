```python
from selenium import webdriver
import time
from time import sleep
import datetime

driver = webdriver.Chrome()
driver.set_window_size(1024, 1200)


now_date = datetime.datetime.today().strftime("%Y%m%d%H")
filepath_sd = r'X:xxx/'
filepath_usb = r'X:xxx/'
filepath_vc = r'X:xxx/'

#SDメモリーカード デイリー
driver.get('https://ranking.rakuten.co.jp/daily/213302/')
time.sleep(5)
driver.save_screenshot( filepath_sd +'sd_daily' + now_date + '.png')
#SDメモリーカード 週間
driver.get('https://ranking.rakuten.co.jp/weekly/213302/')
time.sleep(5)
driver.save_screenshot(filepath_sd +'sd_weekly' + now_date + '.png')

#USBデイリー
driver.get('https://ranking.rakuten.co.jp/daily/303078/')
time.sleep(5)
driver.save_screenshot( filepath_usb +'usb_daily' + now_date + '.png')
#USB週間
driver.get('https://ranking.rakuten.co.jp/weekly/303078/')
time.sleep(5)
driver.save_screenshot( filepath_usb +'usb_weekly' + now_date + '.png')

#ビデオキャプチャデイリー
driver.get('https://ranking.rakuten.co.jp/daily/408533/')
time.sleep(5)
driver.save_screenshot( filepath_vc +'vc_daily' + now_date + '.png')
#ビデオキャプチャ週間
driver.get('https://ranking.rakuten.co.jp/weekly/408533/')
time.sleep(5)
driver.save_screenshot( filepath_vc +'vc_weekly' + now_date + '.png')

driver.quit()


```
